```python
from pathlib import Path

project_path = Path("..")
dataset_path = project_path / "dataset"

print("Project folder:", project_path.resolve())
print("Dataset folder exists:", dataset_path.exists())
```

```python
dataset_files = list(dataset_path.iterdir())

print("Files inside dataset folder:")
for file in dataset_files:
    print(file.name)
```

```python
import pandas as pd

questions_path = dataset_path / "sample_questions.csv"
questions_df = pd.read_csv(questions_path)

print(questions_df.head())
```

```python
answers = {
    "What is Artificial Intelligence?":
        "Artificial Intelligence enables computers to perform tasks that normally require human intelligence.",

    "Explain Python programming.":
        "Python is a programming language used to develop applications, automate tasks, and analyze data.",

    "What is Machine Learning?":
        "Machine Learning is a branch of AI that helps computers learn patterns from data.",

    "How does a computer network work?":
        "A computer network connects devices so they can communicate and share information.",

    "What are the benefits of learning coding?":
        "Learning coding improves problem-solving skills and helps us build software applications."
}

print("Answers prepared successfully!")
print("Total questions:", len(answers))
```

```python
for question, answer in answers.items():
    print("Question:", question)
    print("Answer:", answer)
    print("-" * 60)
```

```python
def chatbot():
    print("Welcome to the AI Chatbot!")
    print("Type 'exit' to stop.")

    while True:
        user_question = input("You: ").strip()

        if user_question.lower() == "exit":
            print("Chatbot: Goodbye!")
            break

        answer = answers.get(user_question)

        if answer:
            print("Chatbot:", answer)
        else:
            print("Chatbot: Sorry, I don't know that answer yet.")

chatbot()
```
